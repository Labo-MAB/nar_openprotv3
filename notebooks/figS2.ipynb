### General statistics per species for OpenProt v3.0

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

### Predictions

```python
def plot_predictions(df, species_sn):
    
    df = df[df["species_sn"]==species_sn]
    labels = df["labels"].to_list()
    values = df["values"].to_list()
    colors = df["colors"].to_list()

    fig, ax = plt.subplots(figsize=(4,4))

    wedges, texts, autotexts = ax.pie(
        values,
        labels=labels,
        colors=colors,
        startangle=0,
        counterclock=True,
        autopct="%.1f%%",
        pctdistance=0.62,
        labeldistance=1.12,
        textprops={"fontsize": 12},
        #radius = 1,
    )

    for autotext in autotexts:
        autotext.set_color("white")

    ax.set_title("Predictions", fontsize=12)
    ax.axis("equal")

    # Fixed axes position
    ax.set_position([0.05, 0.05, 0.90, 0.85])

    #formats = ["svg", "png"]
    #for format in formats:
    #    save_plot_path = os.path.join(save_plots_dir, species_sn, f"predictions.{format}")
    #    plt.savefig(save_plot_path, format=format, bbox_inches = 'tight')

    plt.show()
```

```python
predictions_df = pd.read_csv("../input_data/predictions.csv")
```

```python
for species_sn in predictions_df["species_sn"].unique():
    print("Processing ", species_sn)
    plot_predictions(predictions_df, species_sn)
```

#### AltProts detection

```python
def plot_altprots_detection(df, species_sn):

    df = df[df["species_sn"]==species_sn]
    labels = df["labels"].to_list()
    values = df["values"].to_list()
    colors = df["colors"].to_list()

    # remove zeros
    filtered = [(v, l, c) for v, l, c in zip(values, labels, colors) if v != 0]
    values, labels, colors = map(list, zip(*filtered)) if filtered else ([], [], [])

    fig, ax = plt.subplots(figsize=(4,4))

    wedges, texts, autotexts = ax.pie(
        values,
        labels=labels,
        colors=colors,
        startangle=0,
        counterclock=True,
        autopct="%.1f%%",
        pctdistance=0.62,
        labeldistance=1.12,
        textprops={"fontsize": 12}
    )

    ax.set_title("AltProts detection", fontsize=12)
    ax.axis("equal")

    ax.set_position([0.05, 0.05, 0.90, 0.85])

    #formats = ["svg", "png"]
    #for format in formats:
    #    save_plot_path = os.path.join(save_plots_dir, species_sn, f"altprots_detection.{format}")
    #    plt.savefig(save_plot_path, format=format, bbox_inches = 'tight')

    plt.show()
```

```python
altdetect_df = pd.read_csv("../../nar_openprotv3/input_data/altdetect.csv")
```

```python
for species_sn in altdetect_df["species_sn"].unique():
    print("Processing ", species_sn)
    plot_altprots_detection(altdetect_df, species_sn)
```

#### Transcript localization of detected AltProts

```python
def plot_loc_altprots_detected(
    df,
    species_sn,
):

    figsize=(4,4)
    title="Transcript localization of detected AltProts"
    
    df = df[df["species_sn"]==species_sn]
    methods = df[df["circle"]=="inner"]["labels"].to_list()
    positions = df[df["circle"]=="outer"].drop_duplicates("labels", keep="first")["labels"].to_list()
    inner_values = df[df["circle"]=="inner"]["values"].to_list()
    inner_colors = df[df["circle"]=="inner"]["colors"].to_list()
    outer_values = df[df["circle"]=="outer"]["values"].to_list()
    outer_colors = df[df["circle"]=="outer"]["colors"].to_list()



    # -------------------------
    # Plot
    # -------------------------
    fig, ax = plt.subplots(figsize=figsize)

    # Inner ring
    # width == radius -> no hole in the middle
    wedges_inner, _ = ax.pie(
        inner_values,
        radius=0.70,
        colors=inner_colors,
        startangle=140,
        counterclock=False,
        wedgeprops=dict(width=0.70, edgecolor="white"),
    )

    # Outer ring
    wedges_outer, _ = ax.pie(
        outer_values,
        radius=1.0,
        colors=outer_colors,
        startangle=140,
        counterclock=False,
        wedgeprops=dict(width=0.30, edgecolor="white"),
    )

    # -------------------------
    # Inner labels
    # -------------------------
    for wedge, pos, value in zip(wedges_inner, positions, inner_values):

        angle = (wedge.theta1 + wedge.theta2) / 2
        x = 0.5 * np.cos(np.deg2rad(angle))
        y = 0.5 * np.sin(np.deg2rad(angle))

        ax.text(
            x,
            y + 0.04,
            f"{pos}",
            ha="center",
            va="center",
            fontsize=12,
        )

        ax.text(
            x,
            y - 0.07,
            f"({value})",
            ha="center",
            va="center",
            fontsize=12,
        )

    # -------------------------
    # Outer labels
    # -------------------------
    for wedge, pos, method, value in zip(
        wedges_outer,
        [p for p in positions for _ in methods],
        [m for _ in positions for m in methods],
        outer_values,
    ):

        # Don't label empty slices
        if value == 0:
            continue

        angle = (wedge.theta1 + wedge.theta2) / 2
        r = 0.9

        x = r * __import__("numpy").cos(__import__("numpy").deg2rad(angle))
        y = r * __import__("numpy").sin(__import__("numpy").deg2rad(angle))

        ax.text(
            x,
            y,
            f"{method} : ({value})",
            ha="left" if x >= 0 else "right",
            va="center",
            fontsize=12,
        )

    ax.set_title(title, fontsize=12)
    ax.axis("equal")

    ax.set_position([0.05, 0.05, 0.90, 0.85])

    #formats = ["svg", "png"]
    #for format in formats:
    #    save_plot_path = os.path.join(save_plots_dir, species_sn, f"loc_altprots_detected.{format}")
    #    plt.savefig(save_plot_path, format=format, bbox_inches = 'tight')

    plt.show()
```

```python
localtdetect_df = pd.read_csv("../input_data/loc_altdetect.csv")
```

```python
for species_sn in altdetect_df["species_sn"].unique():
    print("Processing ", species_sn)
    plot_loc_altprots_detected(localtdetect_df, species_sn)
```
